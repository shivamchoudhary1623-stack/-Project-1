```python
def reverse_number(n):
    reverse = 0
    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10
    return reverse
n = int(input("enter a number: "))
print("reverse =", reverse_number(n))
```

```python
def count_numbers(numbers):
  positive = 0
  negative = 0
  zero = 0
  for n in numbers:
    if n > 0:
      positive = positive + 1
    elif n < 0:
      negative = negative + 1
    else:
        zero = zero + 1
  return positive, negative, zero
numbers = [10, -5, 0, 8, -2, 0, 7]
positive, negative, zero = count_numbers(numbers)
print("Positive numbers =", positive)
print("Negative numbers =", negative)
print("Zeros =", zero)
```

```python
def get_units():
    units = int(input("enter electricity units :"))
    return units
def calculate_bill(units):
    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = 100 * 5 + (units - 100) * 7
    else:
        bill = 100 * 5 + 100 * 7 + (units - 200) * 10
    return bill
def display_bill(units, bill):
    print("units consumed =", units)
    print("electricity bill =", bill)
units = get_units()
bill = calculate_bill(units)
display_bill(units, bill)
```

```python
def input_items():
    items = []
    prices = []
    quantities = []
    n = int(input("enter a number of itmes: "))
    for i in range(n):
        item = input("enter item name: ")
        price = float(input("enter price: "))
        quantity = int(input("enter quantity: "))
        items.append(item)
        prices.append(price)
        quantities.append(quantity)
    return items, prices, quantities
def calculate_total(prices, quantities):
    total = 0
    for i in range(len(prices)):
        total = total + prices[i] * quantities[i]
    return total
def calculate_discount(total):
    if total >= 1000:
        discount = total * 10 / 100
    elif total >= 500:
        discount = total * 5 / 100
    else:
        discount = 0
    return discount
def generate_bill(items, prices, quantities, total, discount):
    print("\n-----Bill -----")
    for i in range(len(items)):
        amount = prices[i] * quantities[i]
        print(items[i], "=", amount)
    final_amount = total - discount
    print("Total =", total)
    print("Discount =",discount)
    print("Final Amount =", final_amount)
items, prices, quantities = input_items()
total = calculate_total(prices, quantities)
discount = calculate_discount(total)
generate_bill(items, prices, quantities, total, discount)
```
